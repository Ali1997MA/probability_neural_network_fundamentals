- Calculate the probability of events

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

```python
# data of activities
activity_data = ['walking', 'running', 'cycling', 'swimming', 'walking', 'running', 'cycling', 'walking']
activity_names = {0: 'walking', 1: 'running', 2: 'cycling', 3: 'swimming'}
```

```python
# Function to calculate the probability of an activity in a given sequence

def calculate_probability(sequence):

    counts = Counter(sequence)
    total_count = len(sequence)

    probabilities = {activity: count / total_count for activity, count in counts.items()}
    return probabilities
```

```python
# Calculate the probabilities of activities in the given data
activity_probabilities = calculate_probability(activity_data)
print("Activity Probabilities:", activity_probabilities)
```

```python
for activity, probability in sorted(activity_probabilities.items()):
    print(f"Probability of {activity}: {probability:.2f}")
```

```python
# Print the activity probabilities
for activity, probability in sorted(activity_probabilities.items()):
    print(f"Probability of {activity}: {probability:.2%}")

# Visualize the probability distribution
activities = sorted(activity_probabilities)
probabilities = [activity_probabilities[activity] for activity in activities]

plt.figure(figsize=(10, 5))
plt.bar(activities, probabilities, color="steelblue")
plt.title("Probability of Each Activity")
plt.xlabel("Activity")
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
